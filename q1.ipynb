```python
from IPython.display import display

def run_data_console(data):
    working_df = data.copy()

    while True:
        print("\n===== DATA MANIPULATION SYSTEM =====")
        print("1. Show full table")
        print("2. Show column names")
        print("3. Filter rows (column = value)")
        print("4. Search rows (column contains keyword)")
        print("5. Update a value (row index, column, new value)")
        print("6. Show top N rows sorted by column")
        print("7. Group summary (count by column)")
        print("8. Exit")
        choice = input("Choose an option (1-8): ").strip()

        if choice == '1':
            display(working_df)

        elif choice == '2':
            print(list(working_df.columns))

        elif choice == '3':
            col = input("Column name: ").strip()
            val = input("Value to match: ").strip()
            if col in working_df.columns:
                result = working_df[working_df[col].astype(str) == val]
                print(f"{len(result)} row(s) found.")
                display(result)
            else:
                print("Invalid column name.")

        elif choice == '4':
            col = input("Column name: ").strip()
            keyword = input("Keyword: ").strip()
            if col in working_df.columns:
                result = working_df[working_df[col].astype(str).str.contains(keyword, case=False, na=False)]
                print(f"{len(result)} row(s) found.")
                display(result)
            else:
                print("Invalid column name.")

        elif choice == '5':
            col = input("Column name to update: ").strip()
            if col not in working_df.columns:
                print("Invalid column name.")
                continue
            try:
                idx = int(input("Row index: ").strip())
            except ValueError:
                print("Row index must be a number.")
                continue
            if idx not in working_df.index:
                print("Row index not found.")
                continue
            new_val = input("New value: ").strip()
            working_df.at[idx, col] = new_val
            print("Updated row:")
            display(working_df.loc[[idx]])

        elif choice == '6':
            col = input("Column name to sort by: ").strip()
            n_raw = input("How many rows: ").strip()
            if col in working_df.columns and n_raw.isdigit():
                display(working_df.sort_values(by=col, ascending=False).head(int(n_raw)))
            else:
                print("Invalid column or number.")

        elif choice == '7':
            col = input("Column name to group by: ").strip()
            if col in working_df.columns:
                print(working_df.groupby(col).size().sort_values(ascending=False))
            else:
                print("Invalid column name.")

        elif choice == '8':
            print("Exiting data system. Goodbye!")
            break

        else:
            print("Invalid choice, try again.")

    return working_df


df = run_data_console(df)
```
