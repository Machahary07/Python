```python
# Linear Search
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Linear Search checks each element one by one till target found.
# linear-search(a, n, key)
# 1. for i ← 0 to n-1 do
# 2.     if (a[i] == key) then
# 3.         return i
# 4. return -1
# 
# Complexity Analysis:
# T(n) = O(n), S(n) = O(1)
# 
# Mathematically:
# Best = 1 comparison, Worst = n comparisons
# T(n) = n ⇒ O(n)
# 
# Example: a = [46, 91, 33, 2, 69], key = 2
# i=0: 46≠2 → i=1: 91≠2 → i=2: 33≠2 → i=3: 2=2 ✓ return 3
# (4 comparisons)

if __name__ == "__main__":
    arr = [46, 91, 33, 2, 69]
    target = 2
    result = linear_search(arr, target)
    if result != -1:
        print(f"Element found at index: {result}")
    else:
        print("Element not found in the array.")
```

```python
# Binary Search (sorted arr)
def binary_search(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

# Binary Search
# Array must be sorted. Compares middle, halves search space each step.
# 
# binary-search(a, n, key)
# 1. low ← 0, high ← n-1
# 2. while (low ≤ high) do
# 3.     mid ← (low + high)/2
# 4.     if (a[mid] == key) then
# 5.         return mid
# 6.     else if (a[mid] < key) then
# 7.         low ← mid + 1
# 8.     else
# 9.         high ← mid - 1
# 10. return -1
# 
# complexity analysis:
# T(n) = O(log n), S(n) = O(1)
# 
# Mathematically:
# n → n/2 → n/4 → ... → 1
# n/2ᵏ = 1 ⇒ k = log₂n
# ∴ T(n) ≈ O(log n)
# 
# Example: a = [2, 33, 46, 69, 91], key = 69
# Step 1: low=0, high=4, mid=2 → a[2]=46 < 69 → low=3
# Step 2: low=3, high=4, mid=3 → a[3]=69 ✓ return 3
# (2 comparisons)

if __name__ == "__main__":
    sorted_arr = [2, 33, 46, 69, 91]
    target = 69
    result = binary_search(sorted_arr, target)
    if result != -1:
        print(f"Element found at index: {result}")
    else:
        print("Element not found in the array.")
```

```python
# Selection Sort
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_i = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_i]:
                min_i = j
        arr[i], arr[min_i] = arr[min_i], arr[i]
    return arr

# Selection Sort
# Selects smallest element, places at correct position each pass.
# 
# selection-sort(a, n)
# 1. for i ← 0 to n-2 do
# 2.     min ← i
# 3.     for j ← i+1 to n-1 do
# 4.         if (a[j] < a[min]) then
# 5.             min ← j
# 6.     exchange (a[i], a[min])
#
# Complexity:
# T(n) = O(n²), S(n) = O(1)
# 
# Mathematically:
# T(n) = (n-1) + (n-2) + ... + 2 + 1
# = n(n-1)/2
# = ½(n² − n)
# ∴ T(n) ≈ O(n²)
# 
# Example: a = [46, 91, 33, 2, 69], n=5
# Pass 1: min=2 → swap → [2, 91, 33, 46, 69]
# Pass 2: min=33 → swap → [2, 33, 91, 46, 69]
# Pass 3: min=46 → swap → [2, 33, 46, 91, 69]
# Pass 4: min=69 → swap → [2, 33, 46, 69, 91] ✓ sorted
# (n=5, comparisons = 10)

if __name__ == "__main__":
    arr = [46, 91, 33, 2, 69]
    sorted_arr = selection_sort(arr)
    print(f"Sorted array: {sorted_arr}")
```

```python
# Bucket Sort
# Distributes elements into buckets, sorts each bucket, concatenates.

# bucket-sort(a, n)
# 1. create n empty buckets
# 2. for i ← 0 to n-1 do
# 3.     index ← n × a[i]        // for a[i] in [0,1)
# 4.     insert a[i] into bucket[index]
# 5. for i ← 0 to n-1 do
# 6.     sort bucket[i]           // insertion sort
# 7. concatenate all buckets into a list
# 
# T(n) = O(n + k) avg, O(n²) worst, S(n) = O(n + k)
# 
# Mathematically:
# Distribution = n, Sorting buckets ≈ n (uniform), Concatenate = n
# T(n) = n + n + n ⇒ O(n)
# Worst: all in one bucket ⇒ O(n²)

def bucket_sort(arr):
    n = len(arr)
    buckets = [[] for _ in range(n)]
    
    # Distribute elements into buckets
    for num in arr:
        index = int(n * num)  # Assuming arr contains values in [0, 1)
        if index == n:  # Handle the case where num is exactly 1
            index -= 1
        buckets[index].append(num)
    
    # Sort each bucket and concatenate
    sorted_arr = []
    for bucket in buckets:
        sorted_arr.extend(sorted(bucket))  # Using built-in sort for simplicity
    
    return sorted_arr

# Example: a = [0.42, 0.91, 0.23, 0.02, 0.69], n=5
# index = 5 × a[i]:
# 0.42→b[2], 0.91→b[4], 0.23→b[1], 0.02→b[0], 0.69→b[3]
# b[0]=[0.02], b[1]=[0.23], b[2]=[0.42], b[3]=[0.69], b[4]=[0.91]
# Concatenate → [0.02, 0.23, 0.42, 0.69, 0.91] ✓

if __name__ == "__main__":
    arr = [0.42, 0.91, 0.23, 0.02, 0.69]
    sorted_arr = bucket_sort(arr)
    print(f"Sorted array: {sorted_arr}")
```

```python
# (v) Radix Sort
# Sorts digit by digit, LSD to MSD, using counting sort per digit.

# radix-sort(a, n)
# 1. max ← largest element in a
# 2. for exp ← 1; max/exp > 0; exp ← exp × 10 do
# 3.     counting-sort(a, n, exp)   // stable, by digit (a[i]/exp) mod 10

# T(n) = O(d × (n + b)), d = digits, b = base(10), S(n) = O(n + b)

# Mathematically:
# Each pass = O(n + b), passes = d
# T(n) = d(n + b) ⇒ O(dn) ≈ O(n) when d constant

def counting_sort(arr, exp):
    n = len(arr)
    output = [0] * n
    count = [0] * 10  # Base 10 for decimal digits

    # Count occurrences of each digit
    for i in range(n):
        index = (arr[i] // exp) % 10
        count[index] += 1

    # Change count[i] so that it contains the actual position of this digit in output[]
    for i in range(1, 10):
        count[i] += count[i - 1]

    # Build the output array
    for i in range(n - 1, -1, -1):
        index = (arr[i] // exp) % 10
        output[count[index] - 1] = arr[i]
        count[index] -= 1

    # Copy the output array to arr[], so that arr[] now contains sorted numbers
    for i in range(n):
        arr[i] = output[i]

def radix_sort(arr):
    # Find the maximum number to know the number of digits
    max_num = max(arr)

    # Do counting sort for every digit
    exp = 1
    while max_num // exp > 0:
        counting_sort(arr, exp)
        exp *= 10

# Example: a = [46, 91, 33, 2, 69], n=5
# Pass 1 (units): 91, 2, 33, 46, 69 → [91, 2, 33, 46, 69]
# Pass 2 (tens): 02, 33, 46, 69, 91 → [2, 33, 46, 69, 91] ✓ sorted
# (d=2 passes)

if __name__ == "__main__":
    arr = [46, 91, 33, 2, 69]
    radix_sort(arr)
    print(f"Sorted array: {arr}")
```

```python
# Insertion Sort
# Picks element, inserts into correct position in sorted left part.

# insertion-sort(a, n)
# 1. for i ← 1 to n-1 do
# 2.     key ← a[i]
# 3.     j ← i - 1
# 4.     while (j ≥ 0 and a[j] > key) do
# 5.         a[j+1] ← a[j]
# 6.         j ← j - 1
# 7.     a[j+1] ← key

# T(n) = O(n²) worst, O(n) best (sorted), S(n) = O(1)

# Mathematically:
# T(n) = 1 + 2 + ... + (n-1)
# = n(n-1)/2
# = ½(n² − n)
# ∴ T(n) ≈ O(n²)

def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Example: a = [46, 91, 33, 2, 69], n=5
# i=1: key=91, 91>46 stays → [46, 91, 33, 2, 69]
# i=2: key=33, insert before 46 → [33, 46, 91, 2, 69]
# i=3: key=2, insert at front → [2, 33, 46, 91, 69]
# i=4: key=69, insert before 91 → [2, 33, 46, 69, 91] ✓ sorted

if __name__ == "__main__":
    arr = [46, 91, 33, 2, 69]
    sorted_arr = insertion_sort(arr)
    print(f"Sorted array: {sorted_arr}")
```
