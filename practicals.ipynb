```python
# Write a Python program to import and export datasets in CSV, Excel, and JSON formats using the Pandas library. 
import pandas as pd

df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie', 'David'],
                    'Age': [25, 30, 35, 40]})

df.to_csv('data.csv', index=False)
df.to_excel('data.xlsx', index=False) #requires openpyxl library to be installed, pip install openpyxl
df.to_json('data.json', orient='records')

df_csv = pd.read_csv('data.csv')
df_excel = pd.read_excel('data.xlsx')
df_json = pd.read_json('data.json')

print("CSV Data:\n", df_csv)

# Notes
# 1. The program creates a DataFrame with sample data and exports it to CSV, Excel, and JSON formats.
# 2. It then imports the data back from these formats and prints the CSV data to the console.
# 3. The openpyxl library is required for exporting to Excel format. You can install it using pip if you haven't already.
# 4. Here pandas is used to handle the data import and export operations, making it easy to work with different file formats.
# 5. Here index = false is used to avoid writing row indices to the output files.
# 6. The 'orient' parameter in the to_json method specifies the format of the JSON output. In this case, 'records' means that each row will be represented as a dictionary in a list.

# Conclusion
# The DataFrame round-tripped through CSV, Excel, and JSON without any loss of data, confirming that Pandas can move data between these formats interchangeably. This makes Pandas a reliable choice for interoperating with tools and pipelines that expect different file formats.
```

```python
# Write a Python program to perform data preprocessing by handling missing values, duplicate records, and inconsistent data.  
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'bob', 'Charlie', 'charlie', 'David', np.nan],
    'Age': [25, 30, np.nan, 35, 35, 40],
    'City': [' New York', 'los angeles ', 'Chicago', 'Chicago', np.nan, 'Boston']
}
df = pd.DataFrame(data)
print("Original Data:\n", df)

# Handling missing values
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Name'] = df['Name'].fillna('Unknown')
df['City'] = df['City'].fillna('Unknown')

# Handling inconsistent data (extra spaces, inconsistent casing)
df['Name'] = df['Name'].str.strip().str.title()
df['City'] = df['City'].str.strip().str.title()

# Handling duplicate records
df = df.drop_duplicates(subset=['Name', 'City'], keep='first')

print("\nCleaned Data:\n", df)

# Notes
# 1. Missing values in the 'Age' column are filled with the mean age, while missing values in 'Name' and 'City' are filled with 'Unknown'.
# 2. The str.strip() method removes leading/trailing whitespace and str.title() standardizes the text casing to fix inconsistent data.
# 3. The drop_duplicates method removes duplicate rows based on the 'Name' and 'City' columns, keeping the first occurrence.
# 4. fillna(), str.strip(), str.title(), and drop_duplicates() are all Pandas methods that make cleaning messy real-world data straightforward.
# 5. Data preprocessing is an essential step before analysis since missing, duplicate, or inconsistent data can lead to inaccurate results.

# Conclusion
# The raw dataset started with 6 rows containing a missing Age, a missing Name, a missing City, and a duplicate person recorded as both 'Charlie' and 'charlie'. After preprocessing, the missing Age was replaced with the column mean (33.0), missing text fields were labeled 'Unknown', casing/whitespace inconsistencies were normalized, and the duplicate 'Charlie'/'charlie' rows were merged into one, leaving 5 clean, unique records ready for analysis.
```

```python
# 3. Write a Python program to perform descriptive statistical analysis including mean, median, mode, variance, standard deviation, and covariance.  
import pandas as pd

data = {
    'Math': [78, 85, 92, 88, 76, 95, 89, 84, 90, 78],
    'Science': [82, 88, 79, 91, 85, 87, 90, 86, 88, 84]
}
df = pd.DataFrame(data)
print("Data:\n", df)

mean_vals = df.mean()
median_vals = df.median()
mode_vals = df.mode().iloc[0]
variance_vals = df.var()
std_vals = df.std()
covariance_matrix = df.cov()

print("\nMean:\n", df.mean())
print("\nMedian:\n", df.median())
print("\nMode:\n", df.mode().iloc[0])
print("\nVariance:\n", df.var())
print("\nStandard Deviation:\n", df.std())
print("\nCovariance Matrix:\n", df.cov())

# Notes
# 1. Mean, median, and mode are measures of central tendency; mode can have multiple values per column, so .iloc[0] takes the first one.
# 2. Variance measures how spread out the data is from the mean, while standard deviation is simply the square root of variance, expressed in the same units as the data.
# 3. Covariance measures how two variables vary together; a positive covariance between Math and Science suggests students who score well in one subject tend to score well in the other.
# 4. Pandas' built-in .mean(), .median(), .mode(), .var(), .std(), and .cov() methods operate column-wise by default, making multivariate descriptive statistics straightforward.

# Conclusion
# Math (mean 85.5) and Science (mean 86.0) have very similar average scores, but Math scores are considerably more spread out (std ~6.47, variance ~41.83) than Science scores (std ~3.65, variance ~13.33), meaning student performance in Math is less consistent. The positive covariance (~5.89) between Math and Science indicates the two subjects tend to rise and fall together, i.e. students who score higher in one subject tend to score higher in the other as well.
```
